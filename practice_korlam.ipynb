Given two arrays a[] and b[] of size n and m respectively. The task is to find union between
these two arrays.
Union of the two arrays can be defined as the set containing distinct elements from both the
arrays. If there are repetitions, then only one occurrence of element should be printed in the
union

```python
nm=input().split()
n=int(nm[0])
m=int(nm[1])

a_str=input().split()
a=[]
for i in range(n):
  a.append(int(a_str[i]))
b_str=input().split()
b=[]
for i in range(m):
  b.append(int(b_str[i]))
u=[]
for i in range(n):
  f=False
  for j in range(len(u)):
    if u[j]==a[i]:
      f=True
      break
    if f!=False:
      u.append(a[i])
for i in range(m):
  f=False
  for j in range(len(u)):
    if u[j]==b[i]:
      f=True
      break
  if f!=False:
    u.append(b[i])
print(len(u))
```

Given an unsorted array arr[] of size N having both negative and positive integers. The task is
place all negative element at the end of array without changing the order of positive element
and negative element.
Example 1:
Input :
N = 8
arr = [1, -1, 3, 2, -7, -5, 11, 6]
Output :
1 3 2 11 6 -1 -7 -5
Example 2:
Input :
N=8
arr =[-5, 7, -3, -4, 9, 10, -1, 11]
Output :
7 9 10 11 -5 -3 -4 -1

```python
n=int(input())
arr_str=input().split()
arr=[]
for i in range(n):
 arr.append(int(arr_str[i]))
pos=[]
neg=[]
for i in range(n):
  if arr[i]>=0:
    pos.append(arr[i])
  else:
    neg.append(arr[i])
res=[]
for i in range(len(pos)):
  res.append(pos[i])
for i in range(len(neg)):
  res.append(neg[i])
for i in range(len(res)):
    print(res[i],end=" ")
```
