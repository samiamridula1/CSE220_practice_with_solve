Given an unsorted array arr[] of size N having both negative and positive integers. The task is
place all negative element at the end of array without changing the order of positive element
and negative element.
Example 1:
Input :
N = 8
arr = [1, -1, 3, 2, -7, -5, 11, 6]
Output :
1 3 2 11 6 -1 -7 -5
Example 2:
Input :
N=8
arr =[-5, 7, -3, -4, 9, 10, -1, 11]
Output :
7 9 10 11 -5 -3 -4 -1

```python
n=int(input())
arr_str=input().split()
arr=[]
for i in range(n):
 arr.append(int(arr_str[i]))
pos=[]
neg=[]
for i in range(n):
  if arr[i]>=0:
    pos.append(arr[i])
  else:
    neg.append(arr[i])
res=[]
for i in range(len(pos)):
  res.append(pos[i])
for i in range(len(neg)):
  res.append(neg[i])
for i in range(len(res)):
    print(res[i],end=" ")
```

Given an array of N positive integers and an integer X. The task is to find the frequency of X in
the array.
Example 1:
Input:
N = 5
arr = [1, 1, 1, 1, 1]
X = 1
Output:
5
Explanation: The frequency of 1 is 5.


```python
n1=int(input())
a1=input().split()
a=[]
for i in range(n1):
  a.append(int(a1[i]))
x=int(input())
c=0
for j in range(n1):
  if a[j]==x:
    c+=1
print(f"The frquncy of{x}is {c}")
```

Question 5
Given an array arr and an integer K where K is smaller than size of array, the task is to find the
Kth smallest element in the given array. It is given that all array elements are distinct.
Example 1:
Input:
N = 6
arr = [7, 10, 4, 3, 20, 15]
K = 3
Output : 7
Explanation :
3rd smallest element in the given
array is 7.
Example 2:
Input:
N = 5
arr = [7, 10, 4, 20, 15]
K = 4
Output : 15
Explanation :
4th smallest element in the given
array is 15.

```python
n=int(input())
a=input().split()
arr=[]
for i in range(n):
  arr.append(int(a[i]))
k=3
for i in range(n):
  for j in range(i+1,n):
    if arr[i]>arr[j]:
      tmp=arr[i]
      arr[i]=arr[j]
      arr[j]=tmp
print(f"kth smallest elem is {arr[k-1]}")
```

Given an array arr of N non-negative integers representing the height of blocks. If the width of
each block is 1, compute how much water can be trapped between the blocks during the rainy
season.
Example 1:
Input:
N = 6
arr = [3,0,0,2,0,4]
Output:
10Example 2:
Input:
N = 4
arr = [7,4,0,9]
Output:
10
Explanation:
Water trapped by above
block of height 4 is 3 units and above
block of height 0 is 7 units. So, the
total unit of water trapped is 10 units.
Example 3:
Input:
N = 3
arr = [6,9,9]
Output:
0
Explanation:
No water will be trapped.

```python
# N = 6
# arr = [3,0,0,2,0,4]
N = 3
arr = [6,9,9]
w=0
for i in range(N):
  l_max=0
  for j in range(i+1):
    if arr[j]>l_max:
      l_max=arr[j]
  r_max=0
  for j in range(i,N):
    if arr[j]>r_max:
      r_max=arr[j]
  if l_max<r_max:
    mini=l_max
  else:
    mini=r_max
  w=w+(mini-arr[i])
print(w)
```
